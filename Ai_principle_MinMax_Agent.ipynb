3. Problem Formulation as a Game

          A game in AI consists of:

          Component	Description
          Players	MAX (AI agent) and MIN (Opponent)
          Initial State	Empty game board
          Actions	Legal moves allowed in the game
          Result	New board after a move
          Terminal Test	Win, lose, or draw
          Utility Function	+1 (win), −1 (lose), 0 (draw)

          We use Tic-Tac-Toe as the example game because it clearly demonstrates Minimax.

4. Game Tree Representation

        Each node represents a board state

        Edges represent legal moves

        Leaf nodes are terminal states

        Values are propagated upward using Minimax

            Root (MAX)
           /          \
        MIN            MIN
       /   \          /   \
     MAX   MAX      MAX   MAX


5. The Minimax Algorithm

Minimax assumes:

    MAX tries to maximize the score

    MIN tries to minimize the score

Pseudocode


function MINIMAX(state, isMax):
    if state is terminal:
        return utility(state)

    if isMax:
        best = -∞
        for each move in state:
            best = max(best, MINIMAX(result(move), false))
        return best
    else:
        best = +∞
        for each move in state:
            best = min(best, MINIMAX(result(move), true))
        return best

```python
def check_winner(b):
    wins = [(0,1,2),(3,4,5),(6,7,8),
            (0,3,6),(1,4,7),(2,5,8),
            (0,4,8),(2,4,6)]
    for i,j,k in wins:
        if b[i] == b[j] == b[k] != " ":
            return b[i]
    return None
```

```python
def minimax(board, is_max):
    winner = check_winner(board)
    if winner == "X":
        return 1
    if winner == "O":
        return -1
    if " " not in board:
        return 0

    if is_max:
        best = -float('inf')
        for i in range(9):
            if board[i] == " ":
                board[i] = "X"
                best = max(best, minimax(board, False))
                board[i] = " "
        return best
    else:
        best = float('inf')
        for i in range(9):
            if board[i] == " ":
                board[i] = "O"
                best = min(best, minimax(board, True))
                board[i] = " "
        return best
```

```python
def best_move(board):
    best_val = -float('inf')
    move = -1
    for i in range(9):
        if board[i] == " ":
            board[i] = "X"
            move_val = minimax(board, False)
            board[i] = " "
            if move_val > best_val:
                best_val = move_val
                move = i
    return move
```

```python
def print_board(board):
    print("\n")
    print(" " + board[0] + " | " + board[1] + " | " + board[2] + " ")
    print("-----------")
    print(" " + board[3] + " | " + board[4] + " | " + board[5] + " ")
    print("-----------")
    print(" " + board[6] + " | " + board[7] + " | " + board[8] + " ")
    print("\n")

def play_game_minimax():
    game_board = [" " for _ in range(9)]
    current_player = "X" # AI is 'X', Human is 'O'
    print("Welcome to Tic-Tac-Toe! You are 'O', the AI is 'X'.")

    while True:
        print_board(game_board)
        winner = check_winner(game_board)

        if winner:
            print(f"Player {winner} wins!")
            break
        if " " not in game_board:
            print("It's a draw!")
            break

        if current_player == "X":
            print("AI's turn...")
            move = best_move(game_board)
            game_board[move] = "X"
            current_player = "O"
        else:
            move = int(input("Enter your move (0-8): "))
            if game_board[move] == " ":
                game_board[move] = "O"
                current_player = "X"
            else:
                print("Invalid move. Try again.")

# To play the game, uncomment the line below:
# play_game_minimax()
```

### 7.1 Game Play with Alpha–Beta Pruning

Now, let's create a game play function that utilizes the `best_move_ab` function, which incorporates Alpha–Beta Pruning for potentially faster decision-making.

```python
def alphabeta(board, is_max, alpha, beta):
    winner = check_winner(board)
    if winner == "X":
        return 1
    if winner == "O":
        return -1
    if " " not in board:
        return 0

    if is_max:
        best = -float('inf')
        for i in range(9):
            if board[i] == " ":
                board[i] = "X"
                val = alphabeta(board, False, alpha, beta)
                board[i] = " "
                best = max(best, val)
                alpha = max(alpha, best)
                if beta <= alpha:
                    break  # PRUNE
        return best

    else:
        best = float('inf')
        for i in range(9):
            if board[i] == " ":
                board[i] = "O"
                val = alphabeta(board, True, alpha, beta)
                board[i] = " "
                best = min(best, val)
                beta = min(beta, best)
                if beta <= alpha:
                    break  # PRUNE
        return best
```

```python
def play_game_alphabeta():
    game_board = [" " for _ in range(9)]
    current_player = "X" # AI is 'X', Human is 'O'
    print("Welcome to Tic-Tac-Toe with Alpha-Beta Pruning! You are 'O', the AI is 'X'.")

    while True:
        print_board(game_board)
        winner = check_winner(game_board)

        if winner:
            print(f"Player {winner} wins!")
            break
        if " " not in game_board:
            print("It's a draw!")
            break

        if current_player == "X":
            print("AI (Alpha-Beta) 's turn...")
            move = best_move_ab(game_board)
            game_board[move] = "X"
            current_player = "O"
        else:
            move = int(input("Enter your move (0-8): "))
            if game_board[move] == " ":
                game_board[move] = "O"
                current_player = "X"
            else:
                print("Invalid move. Try again.")

# To play the game, uncomment the line below:
# play_game_alphabeta()
```

```python
def best_move_ab(board):
    best_val = -float('inf')
    move = -1

    for i in range(9):
        if board[i] == " ":
            board[i] = "X"
            move_val = alphabeta(board, False, -float('inf'), float('inf'))
            board[i] = " "
            if move_val > best_val:
                best_val = move_val
                move = i

    return move
```
